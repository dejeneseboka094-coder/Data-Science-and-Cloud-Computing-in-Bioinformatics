### 11: Color PCA by treatment.

```python
# Add treatment labels
def treatment(sample):

    if "Alb_Dex" in sample:
        return "Alb_Dex"

    elif sample.startswith("Alb"):
        return "Alb"

    elif sample.startswith("Dex"):
        return "Dex"

    else:
        return "Untreated"

pca_df["Treatment"] = pca_df["Sample"].apply(treatment)
```
